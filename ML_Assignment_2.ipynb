```python

# STAGE 2 - STEP 1
# Loading the Breast Cancer Dataset

from sklearn.datasets import load_breast_cancer
import pandas as pd

# Load dataset
cancer = load_breast_cancer()

# Convert features into a DataFrame
df = pd.DataFrame(
    cancer.data,
    columns=cancer.feature_names
)

# Add target column
df["diagnosis"] = cancer.target

# Display first 5 records
print(df.head())
```

```python

# Check dataset dimensions

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
print(df.info())
```

```python

# Step 4: Data quality check

print("Dataset shape:", df.shape)

# Check missing values
print("\nMissing values:")
print(df.isnull().sum().sum())

# Check target distribution
print("\nDiagnosis distribution:")
print(df["diagnosis"].value_counts())

# Display diagnosis names
print("\nDiagnosis mapping:")
print(dict(enumerate(cancer.target_names)))
```

```python

# Step 4: Data quality check

print("Dataset shape:", df.shape)

# Check missing values
print("\nMissing values:")
print(df.isnull().sum().sum())

# Check target distribution
print("\nDiagnosis distribution:")
print(df["diagnosis"].value_counts())

# Display diagnosis names
print("\nDiagnosis mapping:")
print(dict(enumerate(cancer.target_names)))
```

```python

# Step 5: Separate features and target

X = df.drop("diagnosis", axis=1)
y = df["diagnosis"]

# Split dataset into training and testing data

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display dataset sizes

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target:", y_train.shape)
print("Testing Target:", y_test.shape)

print("\nTraining category distribution:")
print(y_train.value_counts())

print("\nTesting category distribution:")
print(y_test.value_counts())

```

```python

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Create Logistic Regression model
logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Train the model
logistic_model.fit(X_train, y_train)

print("SUCCESS: Logistic Regression trained!")
```

```python

# Make predictions
y_pred = logistic_model.predict(X_test)

print("Actual:", y_test.values[:10])
print("Predicted:", y_pred[:10])
```

```python

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", round(accuracy * 100, 2), "%")
```

```python

# Logistic Regression - Complete Evaluation

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# Get prediction probabilities for AUC
y_probability = logistic_model.predict_proba(X_test)[:, 1]

# Calculate all six metrics
logistic_results = {
    "Model": "Logistic Regression",
    "Accuracy": accuracy_score(y_test, y_pred),
    "AUC": roc_auc_score(y_test, y_probability),
    "Precision": precision_score(y_test, y_pred),
    "Recall": recall_score(y_test, y_pred),
    "F1": f1_score(y_test, y_pred),
    "MCC": matthews_corrcoef(y_test, y_pred)
}

# Display results
print("LOGISTIC REGRESSION - EVALUATION RESULTS")
print("----------------------------------------")

for metric, value in logistic_results.items():
    if metric != "Model":
        print(metric, ":", round(value, 4))
```

```python

# Store Logistic Regression results

results = [logistic_results]

print("Logistic Regression results saved!")
```

```python

# STAGE 3 - MODEL 2
# Decision Tree Classifier

from sklearn.tree import DecisionTreeClassifier

# Create the model
decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

# Train the model
decision_tree_model.fit(X_train, y_train)

print("SUCCESS: Decision Tree trained!")
```

```python

# Predict using testing data

dt_pred = decision_tree_model.predict(X_test)

print("Actual values:")
print(y_test.values[:10])

print("\nPredicted values:")
print(dt_pred[:10])
```

```python

# Decision Tree - Evaluation

dt_probability = decision_tree_model.predict_proba(X_test)[:, 1]

dt_results = {
    "Model": "Decision Tree",
    "Accuracy": accuracy_score(y_test, dt_pred),
    "AUC": roc_auc_score(y_test, dt_probability),
    "Precision": precision_score(y_test, dt_pred),
    "Recall": recall_score(y_test, dt_pred),
    "F1": f1_score(y_test, dt_pred),
    "MCC": matthews_corrcoef(y_test, dt_pred)
}

print("DECISION TREE - EVALUATION RESULTS")
print("----------------------------------")

for metric, value in dt_results.items():
    if metric != "Model":
        print(metric, ":", round(value, 4))
```

```python

results.append(dt_results)

print("Decision Tree results saved!")
print("Total models evaluated:", len(results))
```

```python

# STAGE 3 - MODEL 3
# K-Nearest Neighbors

from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Create KNN pipeline
knn_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", KNeighborsClassifier(n_neighbors=5))
])

# Train model
knn_model.fit(X_train, y_train)

print("SUCCESS: KNN model trained!")
```

```python

knn_pred = knn_model.predict(X_test)

print("Actual values:")
print(y_test.values[:10])

print("\nPredicted values:")
print(knn_pred[:10])
```

```python

# KNN - Evaluation

knn_probability = knn_model.predict_proba(X_test)[:, 1]

knn_results = {
    "Model": "KNN",
    "Accuracy": accuracy_score(y_test, knn_pred),
    "AUC": roc_auc_score(y_test, knn_probability),
    "Precision": precision_score(y_test, knn_pred),
    "Recall": recall_score(y_test, knn_pred),
    "F1": f1_score(y_test, knn_pred),
    "MCC": matthews_corrcoef(y_test, knn_pred)
}

print("KNN - EVALUATION RESULTS")
print("------------------------")

for metric, value in knn_results.items():
    if metric != "Model":
        print(metric, ":", round(value, 4))
```

```python

results.append(knn_results)

print("KNN results saved!")
print("Total models evaluated:", len(results))
```

```python

# STAGE 3 - MODEL 4
# Gaussian Naive Bayes

from sklearn.naive_bayes import GaussianNB

# Create model
nb_model = GaussianNB()

# Train model
nb_model.fit(X_train, y_train)

print("SUCCESS: Gaussian Naive Bayes trained!")
```

```python

nb_pred = nb_model.predict(X_test)

print("Actual values:")
print(y_test.values[:10])

print("\nPredicted values:")
print(nb_pred[:10])
```

```python

# Gaussian Naive Bayes - Evaluation

nb_probability = nb_model.predict_proba(X_test)[:, 1]

nb_results = {
    "Model": "Gaussian Naive Bayes",
    "Accuracy": accuracy_score(y_test, nb_pred),
    "AUC": roc_auc_score(y_test, nb_probability),
    "Precision": precision_score(y_test, nb_pred),
    "Recall": recall_score(y_test, nb_pred),
    "F1": f1_score(y_test, nb_pred),
    "MCC": matthews_corrcoef(y_test, nb_pred)
}

print("NAIVE BAYES - EVALUATION RESULTS")
print("--------------------------------")

for metric, value in nb_results.items():
    if metric != "Model":
        print(metric, ":", round(value, 4))
```

```python

results.append(nb_results)

print("Naive Bayes results saved!")
print("Total models evaluated:", len(results))
```

```python

# STAGE 3 - MODEL 5
# Support Vector Machine

from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

# Create SVM pipeline
svm_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", SVC(
        kernel="rbf",
        probability=True,
        random_state=42
    ))
])

# Train model
svm_model.fit(X_train, y_train)

print("SUCCESS: SVM model trained!")
```

```python

svm_pred = svm_model.predict(X_test)

print("Actual values:")
print(y_test.values[:10])

print("\nPredicted values:")
print(svm_pred[:10])
```

```python

# SVM - Evaluation

svm_probability = svm_model.predict_proba(X_test)[:, 1]

svm_results = {
    "Model": "Support Vector Machine",
    "Accuracy": accuracy_score(y_test, svm_pred),
    "AUC": roc_auc_score(y_test, svm_probability),
    "Precision": precision_score(y_test, svm_pred),
    "Recall": recall_score(y_test, svm_pred),
    "F1": f1_score(y_test, svm_pred),
    "MCC": matthews_corrcoef(y_test, svm_pred)
}

print("SVM - EVALUATION RESULTS")
print("------------------------")

for metric, value in svm_results.items():
    if metric != "Model":
        print(metric, ":", round(value, 4))
```

```python

results.append(svm_results)

print("SVM results saved!")
print("Total models evaluated:", len(results))
```

```python

# STAGE 4 - MODEL COMPARISON

import pandas as pd

# Convert all model results into a DataFrame
comparison_df = pd.DataFrame(results)

# Display all metrics rounded to 4 decimals
comparison_df = comparison_df.round(4)

print("FINAL MODEL COMPARISON")
print("======================")

display(comparison_df)
```

```python

%%writefile app.py

import streamlit as st
import pandas as pd
import joblib
import os

from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef,
    confusion_matrix,
    classification_report
)

# Page configuration
st.set_page_config(
    page_title="Breast Cancer ML Dashboard",
    page_icon="🔬",
    layout="wide"
)

st.title("🔬 Breast Cancer Classification Dashboard")
st.write(
    "Compare five machine learning models "
    "using the Breast Cancer Wisconsin dataset."
)

# Model paths
model_files = {
    "Logistic Regression": "model/logistic_regression.joblib",
    "Decision Tree": "model/decision_tree.joblib",
    "KNN": "model/knn.joblib",
    "Gaussian Naive Bayes": "model/naive_bayes.joblib",
    "Support Vector Machine": "model/svm.joblib"
}

# Load saved models
@st.cache_resource
def load_models():
    loaded_models = {}

    for name, path in model_files.items():
        loaded_models[name] = joblib.load(path)

    return loaded_models

# Check model files
if not all(os.path.exists(path) for path in model_files.values()):
    st.error("Model files are missing. Please check the model folder.")
    st.stop()

models = load_models()

# Dataset upload
st.header("1. Upload Test Dataset")

uploaded_file = st.file_uploader(
    "Upload your test_data.csv file",
    type=["csv"]
)

if uploaded_file is not None:

    data = pd.read_csv(uploaded_file)

    st.success("Dataset uploaded successfully!")

    st.write("Dataset shape:", data.shape)

    # Target column
    target_column = "diagnosis"

    if target_column not in data.columns:
        st.error("Target column 'diagnosis' is missing.")
        st.stop()

    # Get required feature names
    feature_names = list(models["Logistic Regression"].feature_names_in_)

    missing_features = [
        col for col in feature_names if col not in data.columns
    ]

    if missing_features:
        st.error(f"Missing features: {missing_features}")
        st.stop()

    X = data[feature_names]
    y_true = data[target_column]

    if not set(y_true.unique()).issubset({0, 1}):
        st.error("Diagnosis must contain only 0 and 1.")
        st.stop()

    st.subheader("Dataset Preview")
    st.dataframe(data.head())

    # Calculate metrics for all models
    all_results = {}

    for name, model in models.items():

        predictions = model.predict(X)
        probabilities = model.predict_proba(X)[:, 1]

        all_results[name] = {
            "Accuracy": accuracy_score(y_true, predictions),
            "AUC": roc_auc_score(y_true, probabilities),
            "Precision": precision_score(
                y_true, predictions, zero_division=0
            ),
            "Recall": recall_score(
                y_true, predictions, zero_division=0
            ),
            "F1": f1_score(
                y_true, predictions, zero_division=0
            ),
            "MCC": matthews_corrcoef(y_true, predictions)
        }

    # Model selection
    st.header("2. Select Machine Learning Model")

    selected_model = st.selectbox(
        "Choose a model",
        list(models.keys())
    )

    model = models[selected_model]

    y_pred = model.predict(X)
    y_probability = model.predict_proba(X)[:, 1]

    # Evaluation metrics
    st.header("3. Model Evaluation Metrics")

    metrics = all_results[selected_model]

    columns = st.columns(3)

    for col, (metric, value) in zip(columns + st.columns(3), metrics.items()):
        col.metric(metric, f"{value:.4f}")

    # Confusion matrix
    st.header("4. Confusion Matrix")

    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])

    cm_df = pd.DataFrame(
        cm,
        index=["Actual Malignant", "Actual Benign"],
        columns=["Predicted Malignant", "Predicted Benign"]
    )

    st.dataframe(cm_df)

    # Classification report
    st.header("5. Classification Report")

    report = classification_report(
        y_true,
        y_pred,
        target_names=["Malignant", "Benign"],
        output_dict=True,
        zero_division=0
    )

    st.dataframe(pd.DataFrame(report).transpose())

    # Compare all five models
    st.header("6. Comparison of All Five Models")

    comparison = pd.DataFrame(all_results).transpose()

    st.dataframe(
        comparison.style.format("{:.4f}"),
        use_container_width=True
    )

    st.success("Evaluation completed successfully!")

else:
    st.info("Please upload your test_data.csv file to begin.")
```
